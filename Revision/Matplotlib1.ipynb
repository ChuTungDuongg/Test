```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
N_A_x = np.arange(1,9)
```

```python
N_A_x
```

```python
N_A_y = np.arange(3,27,3)
```

```python
N_A_y
```

```python
plt.plot(N_A_x, N_A_y, color = "orange")
plt.show()
```

```python
plt.xlabel("N_A_x")
plt.xlabel("N_A_y")
plt.title("PLOT")
plt.plot(N_A_x,N_A_y)
plt.show()
```

```python
LS_X = np.linspace(-4, 4, 50)
LS_Y = LS_X ** 2
plt.plot(LS_X, LS_Y)   
plt.show()
```

```python
plt.plot(LS_X, LS_Y, ".")
plt.show()
```

```python
plt.plot(LS_X, LS_Y, "r.")
plt.show()
```

```python
plt.plot(LS_X, np.sin(LS_X), "r.")
plt.plot(LS_X, np.cos(LS_X), "g.")
plt.plot(LS_X, np.tanh(LS_X), "b.")
plt.show()
```
